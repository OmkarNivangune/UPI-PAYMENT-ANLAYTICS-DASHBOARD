```python
text_cols = [
    "transaction_type",
    "transaction_status",
    "failure_reason",
    "failure_code",
    "sender_bank",
    "receiver_bank",
    "upi_app",
    "sender_state",
    "device_type",
    "network_type",
    "merchant_category"
]

for col in text_cols:
    df[col] = df[col].astype("string").str.strip()
```

```python
df.columns
```

```python
title_cols = [
    "transaction_type",
    "sender_bank",
    "receiver_bank",
    "upi_app",
    "sender_state",
    "device_type",
    "network_type",
    "merchant_category"
]

for col in title_cols:
    df[col] = df[col].str.title()

df["transaction_status"] = df["transaction_status"].str.upper()
df["failure_code"] = df["failure_code"].str.upper()
df["failure_reason"] = df["failure_reason"].str.upper()
```

```python
df["amount_inr"] = (
    df["amount_inr"]
    .astype("string")
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)
print(df["amount_inr"].head())
```

```python
df["transaction_timestamp"] = pd.to_datetime(
    df["transaction_timestamp"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print(df["transaction_timestamp"].head())
print(df["transaction_timestamp"].isnull().sum())
```

```python
print(df.isnull().sum())
```

```python
df["response_time_ms"] = pd.to_numeric(
    df["response_time_ms"], errors="coerce"
)

df["retry_attempt"] = pd.to_numeric(
    df["retry_attempt"], errors="coerce"
)

print(df[["response_time_ms", "retry_attempt"]].dtypes)
```

```python
df.isnull().sum()
```

```python
category_cols = [
    "failure_reason",
    "failure_code",
    "merchant_category"
]

for col in category_cols:
    df[col] = df[col].fillna("Unknown")
```

```python
df["is_success"] = (
    df["transaction_status"] == "SUCCESS"
).astype(int)

df["is_failed"] = (
    df["transaction_status"] == "FAILED"
).astype(int)

df["is_pending"] = (
    df["transaction_status"] == "PENDING"
).astype(int)
```

```python
print(df["transaction_status"].value_counts())
```

```python
failed_df = df[
    df["transaction_status"] == "FAILED"
]

print(failed_df["failure_reason"].value_counts())
```

```python
print(
    failed_df["upi_app"].value_counts()
)
```

```python
print(df.isnull().sum())
```

```python
total = len(df)

success = (df["transaction_status"] == "SUCCESS").sum()
failed = (df["transaction_status"] == "FAILED").sum()
pending = (df["transaction_status"] == "PENDING").sum()

success_rate = success / total * 100 if total else 0
failure_rate = failed / total * 100 if total else 0

print("Total Transactions:", total)
print("Successful Transactions:", success)
print("Failed Transactions:", failed)
print("Pending Transactions:", pending)
print("Success Rate:", round(success_rate, 2), "%")
print("Failure Rate:", round(failure_rate, 2), "%")
```
